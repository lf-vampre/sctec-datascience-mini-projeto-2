````python
# IMPORTS

import pandas as pd

# Leitura dos CSV's baixados do site do Ministério da Saúde
# Iteração com for dentro do range de 2020 a 2026
# Armazenado em um dicionário chamado dfs

dfs = {ano: pd.read_csv(f"dados/bronze/{ano}.csv", sep=";", encoding="utf-8")
       for ano in range(2020, 2027)}


print('Bibliotecas e arquivos .csv importados.')
````

---

```
┌───────────────────────────────────────────────────────────────────────┐
│ 1- EXTRACT - Verificação e conhecimento da base                       │
│   -> df.head(3) / df.tail(3)                                          │
│   -> df.info()                                                        │
│   -> df.isnull().sum()                                                │
│   -> df.nunique()                                                     │
└───────────────────────────────────────────────────────────────────────┘
```

````python
print(' -------- Exibindo as primeiras 3 linhas da base de dados de cada ano. -------- ')
for ano in range(2020, 2027):
    print(f"\n -------- Ano {ano} --------")
    print(dfs[ano].head(2).to_string())
````

````python
print(' -------- Exibindo as últimas 3 linhas da base de dados de cada ano. -------- ')
for ano in range(2020, 2027):
    print(f"\n -------- Ano {ano} --------")
    print(dfs[ano].tail(2).to_string())
````

````python
print(' -------- Exibindo número de registros, colunas, valores não nulos e tipo de dados. -------- ')
for ano in range(2020, 2027):
    print(f"\n -------- Ano {ano} --------")
    print(dfs[ano].info())
````
